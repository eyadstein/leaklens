```python
import pandas as pd

df = pd.DataFrame({
    'Ism': ['Ali', 'Vali', 'Hasan', 'Husan', 'Aziz'],
    'Shahar': ['Toshkent', 'Samarqand', 'Toshkent', 'Farg‘ona', 'Samarqand'],
    'Yosh': [25, 30, 28, 35, 29]
})

print(df)

guruhlangan = df.groupby('Shahar')['Yosh'].mean()
print(guruhlangan)
```

```python
natija = df.groupby('Shahar')['Yosh'].agg(['min', 'max', 'mean'])
print(natija)
```

```python
pivot = df.pivot_table(values='Yosh', index='Shahar', aggfunc=['mean', 'count'])
print(pivot)
```

```python
df['Guruh_o‘rtacha_yoshi'] = df.groupby('Shahar')['Yosh'].transform('mean')
print(df)
```

```python
import pandas as pd

data = {
    "Ism": ["Ali", "Vali", "Hasan", "Husan", "Aziz"],
    "Kasb": ["Dasturchi", "Dizayner", "Muhandis", "Dasturchi", "Dizayner"],
    "Maosh": [5000, 4500, 5500, 6000, 4700],
    "Yosh": [25, 30, 28, 35, 29]
}

df = pd.DataFrame(data)
```

```python
df['Yosh toifa'] = df['Yosh'].apply(lambda yosh: "Yosh" if yosh < 28 else "O‘rta yosh" if yosh < 35 else "Katta yosh")

df
```
